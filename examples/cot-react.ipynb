## Define your tools

These are functions that your agent can "act" upon.

```python
def get_market_data(ticker: str) -> str:
    """Fetches current trading activity, volume, and recent price movements for a given ticker."""
    # In a real scenario, this would call a financial API (e.g., Alpha Vantage, Polygon.io)
    return f"Market data for {ticker}: (simulate API call: volume=10M, price=15.00, recent_move=+2%)"

def research_news_and_reports(query: str) -> str:
    """Searches for recent news, earnings reports, and significant announcements related to the query."""
    # Simulate a web search or news API call
    return f"News for '{query}': (simulate: Palantir announced new government contract, earnings beat)"

def get_analyst_ratings(ticker: str) -> str:
    """Retrieves analyst ratings, price targets, and institutional sentiment for a given ticker."""
    # Simulate an analyst data API
    return f"Analyst data for {ticker}: (simulate: Buy rating, target $20, institutional sentiment positive)"

def perform_technical_analysis(data: str) -> str:
    """Performs technical analysis including RSI, MACD, Moving Averages based on provided data."""
    # This would involve more complex financial libraries like pandas, TA-Lib
    return f"Technical analysis results for {data}: (simulate: RSI=60, MACD bullish cross, MA shows uptrend)"

def create_visualization_package(analysis_results: str) -> str:
    """Generates a visualization package including candlestick charts, volume plots, and trend lines."""
    # This would involve matplotlib, plotly, or similar libraries
    return f"Visualization package created: (simulate: candlestick chart PNG, volume plot PNG)"

tools = [
    Tool(
        name="get_market_data",
        func=get_market_data,
        description="Useful for getting real-time market data for a stock ticker."
    ),
    Tool(
        name="research_news_and_reports",
        func=research_news_and_reports,
        description="Useful for searching for recent news, earnings, and announcements."
    ),
    Tool(
        name="get_analyst_ratings",
        func=get_analyst_ratings,
        description="Useful for retrieving analyst ratings and sentiment."
    ),
    Tool(
        name="perform_technical_analysis",
        func=perform_technical_analysis,
        description="Useful for computing technical indicators and chart patterns."
    ),
    Tool(
        name="create_visualization_package",
        func=create_visualization_package,
        description="Useful for generating visual charts and plots from analysis results."
    ),
]
```

## Create the Prompt Template for CoT + ReAct

This prompt guides the LLM through the thought process and tool usage.

The `agent_scratchpad` is crucial for ReAct, as it stores previous thoughts and actions.


```python
prompt_template = PromptTemplate.from_template("""
You are a financial analyst assistant. Your goal is to provide a comprehensive analysis of a stock.
Follow these steps carefully:

**Thinking Stage:**
1.  Understand the user's request: Identify the stock ticker.
2.  Formulate a detailed plan for research and computation.

**Developing a plan Stage:**
1.  **Researching:**
    * Gather current market data (trading activity, volume, price movements).
    * Research recent news, earnings reports, and significant announcements (last 30 days).
    * Collect analyst ratings, price targets, and institutional sentiment.
2.  **Computing:**
    * Analyze technical indicators (RSI, MACD, Moving Averages).
    * Identify support/resistance levels.
    * Analyze volume patterns and trends.
    * Generate price trend analysis across multiple timeframes.
    * Create a visualization package (candlestick chart, volume plot, trend lines).

**Researching Stage (Use tools):**
Use the available tools to gather the necessary data. For each piece of information needed from your plan, decide which tool to use.
When you have gathered all necessary information, move to the Computing Stage.

**Computing Stage (Use tools):**
Process the gathered information using your tools to perform analysis and generate visualizations.
When all computations are done, prepare your final response.

**Responding Stage:**
Synthesize all findings into a clear, concise, and insightful financial analysis report, including key takeaways and visualizations (mentioning they were generated).

You have access to the following tools:
{tools}

Use the following format:

Question: the input question you must answer
Thought: you should always think about what to do
Action: the action to take, should be one of [{tool_names}]
Action Input: the input to the action
Observation: the result of the action
... (this Thought/Action/Action Input/Observation can repeat N times)
Thought: I have gathered all necessary information and performed all computations. I can now synthesize the final answer.
Final Answer: the comprehensive financial analysis report

Begin!

Question: {input}
{agent_scratchpad}
""")
```

## Create ReAct Agent

```python
from langchain.agents import AgentExecutor, create_react_agent

agent = create_react_agent(llama_claude_model, tools, prompt_template)
```

## Create Agent Executor

```python
agent_executor = AgentExecutor(agent=agent, tools=tools, verbose=True)
```

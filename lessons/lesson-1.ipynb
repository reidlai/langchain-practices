## Conversation with Model

This is simple example to pass system prompt (SystemMessage) and user prompt (HumanMessage) to LLM model and get response by calling invoke method.  The following example uses bartowski/Meta-Llama-3.1-8B-Claude-GGUF model pulled from Hugging Face (hf.co).

```python
import os

from langchain_ollama import ChatOllama
from langchain_core.language_models.chat_models import BaseChatModel
from langchain_core.messages import HumanMessage, SystemMessage
from dotenv import load_dotenv

load_dotenv()

messages = [
    SystemMessage("Translate the following from English into Italian"),
    HumanMessage("hi!"),
]

model = ChatOllama(model="hf.co/bartowski/Meta-Llama-3.1-8B-Claude-GGUF:Q4_K_M", temperature=0.7, top_k=40)

model.invoke(messages)
```
